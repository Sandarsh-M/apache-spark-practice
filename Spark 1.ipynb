# Apache Spark

## What is Apache Spark?

Apache Spark is an open-source distributed data processing framework used for processing large volumes of data quickly and efficiently. It supports batch processing, streaming, machine learning, graph processing, and SQL analytics.

### Key Features
- Fast in-memory processing
- Distributed computing
- Fault tolerant
- Supports multiple languages (Python, Scala, Java, SQL, R)
- Easy integration with Hadoop, cloud storage, and databases

### Common Use Cases
- ETL Pipelines
- Big Data Processing
- Data Engineering
- Real-time Streaming
- Machine Learning

### Lazy Evaluation

Lazy Evaluation means Spark does not execute transformations immediately. Instead, it builds an execution plan and runs it only when an action (like `show()`, `collect()`, or `count()`) is called.

```python
data= [
    (1,"lisa","lisa@gmail.com"),
    (2,"john","john@gmail.com"),
    (3,"mary","mary@gmail.com"),
    (4,"peter","peter@gmail.com"),
    (5,"jack","jack@gmail.com"),
    (6,"rose","rose@gmail.com")
]

columns=["id","name","email"]

df_new=spark.createDataFrame(data,columns)

#transformation -1
df_new=df_new.select("id","email")

#transformation -2
df_new=df_new.filter(col("id")>5)
```
