## `spark.read.format()`

`spark.read.format()` is used to specify the format of the input data (such as CSV, JSON, Parquet, or Delta) before loading it into a DataFrame.

It creates a DataFrameReader and follows **lazy evaluation**, meaning the data is read only when an action is performed.

```python
df=spark.read.format("csv")\
    .option("header",True)\
        .load(file_path)
```

```python
#Wide Transformation
df=df.repartition(2)

#Narrow Transformation1
df=df.select("order_id","customer_id")

#Narrow Transformation -2
df=df.filter(col("order_id")==1001)

#wide transformation
df=df.groupBy("customer_id").count()
```
