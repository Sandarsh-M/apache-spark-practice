```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
df=spark.read.format("csv")\
        .option("header",True)\
        .option("inferScheam",True)\
        .load("/Volumes/sparkcatalog/raw/source/raw_orders/")
```

```python
display(df)
```

```python
display(df.withColumn("spark_partition",spark_partition_id())\
    .groupBy("spark_partition")\
    .agg(count("*").alias("count"))\
    .orderBy("spark_partition"))
```

```python
df_repart=df.repartition(4)
```

##Query Plans

```python
#This will display all the query plan
df.explain(True)    
```

```python
#This only show the physical plan
df.explain()
```

##Query Plan for repartition

```python
df_repart=df.repartition(4)
df_repart.explain()
```

##Query plan for coalesce

```python
df_coal=df.coalesce(2)
df_coal.explain()
```

##Query plan for aggregate

```python
df_grp=df_coal.withColumn("spark_partition",spark_partition_id())\
    .groupBy("spark_partition")\
    .agg(count("*").alias("count"))\
    .orderBy("spark_partition")

df_grp.explain()
```
