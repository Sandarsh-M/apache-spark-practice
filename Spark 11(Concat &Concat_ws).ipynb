```python
df_csv=spark.read.format("csv")\
            .option("header",True)\
            .option("inferSchema",True)\
            .load("/Volumes/sparkcatalog/raw/source/raw_orders/orders.csv")

display(df_csv)
```

#Concat & Concat_WS

###concat

```python
from pyspark.sql.functions import col,lit,concat

df_con=(df_csv
        .withColumn("custom_id"
        ,concat(col("customer_id"),lit("|")
        ,col("order_id"),lit("|")
        ,col("product_id"))
                    )
)

display(df_con)
```

```python
from pyspark.sql.functions import concat_ws

df_con_ws=(df_csv
           .withColumn("custom_id"
           ,concat_ws(lit("|")
           ,col("customer_id"),col("order_id"),col("product_id"))
            )
)

display(df_con_ws)
```
