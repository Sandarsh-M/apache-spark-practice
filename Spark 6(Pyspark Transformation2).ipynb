##Transformations


###SELECT

```python
df_csv=(spark.read.format("csv")
        .option("header",True)
        .option("inferSchema",True)
        .load("/Volumes/sparkcatalog/raw/source/raw_orders/orders.csv")
        )

display(df_csv)
```

```python
df_select=df_csv.select("order_id","customer_id")

display(df_select)
```

###alias

```python
from pyspark.sql.functions import col
df_alias = df_csv.select(col("order_id").alias("id"), col("customer_id").alias("c_id"))
display(df_alias)
```

###filter

####Scenario 1: Returned Orders

```python
df_filter=df_csv.filter(df_csv.order_status=="Returned")

display(df_filter)
```

####Scenario 2:Returned order or Cancelled Orders

```python
from pyspark.sql.functions import col
df_filter2=df_csv.filter((col("order_status")=="Returned") | (col("order_status")=="Cancelled"))

display(df_filter2)
```

####Scenario 3: Orders -Returned ,Shipped, Cancelled

```python
# df_filter3=df_csv.filter(col("order_status").isin("Returned","Cancelled","Shipped"))
# display(df_filter3)

#Or keep the filter values needed in varailable and add in the script
desired_order=["Returned","Cancelled","Shipped"]
df_filter3=df_csv.filter(col("order_status").isin(desired_order))

display(df_filter3)
```

###withColumnRenamed

```python
# df_renamed=df_csv.withColumnRenamed("order_status","order_status_info")

# display(df_renamed)

df_renamed=df_csv.withColumnRenamed("order_status","order_status_info_new")

display(df_renamed)
```

###WithColumn

####Scenario 1-add new column at the end

```python
from pyspark.sql.functions import lit
df_mod1=df_csv.withColumn("flag",lit("Y"))
display(df_mod1)
```

####Scenario 2-WithColumn correct the shipping address column

```python
from pyspark.sql.functions import regexp_replace
df_mod2=df_csv.withColumn("shipping_address",regexp_replace("shipping_address",",.*",""))

display(df_mod2)
```

####Scenario 3-Additional or multiplication

```python
from pyspark.sql.functions import col,round
df_mod3=df_csv.withColumn("total_price",col("price")*col("quantity"))\
    .withColumn("total_price",round(col("total_price"),2))
display(df_mod3)
```
