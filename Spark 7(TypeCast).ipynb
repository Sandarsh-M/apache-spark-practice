# TypeCasting
###the process of changing a variable or value from one data type to another

```python
df_csv=spark.read.format("csv")\
        .option("header",True)\
        .option("inferSchema",True)\
        .load("/Volumes/sparkcatalog/raw/source/raw_orders/orders.csv")

display(df_csv)
```

```python
from pyspark.sql.functions import col
from pyspark.sql.types import StringType
df_cast1=df_csv.withColumn("order_id",col("order_id").cast(StringType()))

display(df_cast1)
```

```python
df_cast2=df_csv.withColumn("order_id",col("order_id").cast("string"))

display(df_cast2)
```

# Sorting

###Scenario 1: sort order_date column in ascending & descending  order

```python
df_sort1=df_csv.sort(col("order_date").desc())
display(df_sort1)
```

###Scenario 2- Sort order_date descending order and quantity  ascending order

```python
df_sort2=df_csv.sort((col("order_date").desc())\
        ,col("quantity").asc())
display(df_sort2)
```

OR


```python

df_sort2=df_csv.sort(["order_date","quantity"],ascending=[False, True])
display(df_sort2)
```

#Union & UninonBYNamae

###Union

```python
data1=[(1,"aa","123"),(2,"bb","456"),(3,"cc","789")]
df1=spark.createDataFrame(data1,["id","name","address"])

data2=[(4,"aa","123"),(5,"bb","456"),(6,"cc","789"),(7,"dd","123")]
df2=spark.createDataFrame(data2,["id","name",])


df_union=df1.union(df2)
display(df_union)
```

###UnionByName

```python
data1=[(1,"aa","123"),(2,"bb","456"),(3,"cc","789")]
df1=spark.createDataFrame(data1,["id","name","address"])

data2=[("123","aa",4),("456","bb",5),("789","cc",6),("123","dd",7)]
df2=spark.createDataFrame(data2,["address","name","id"])


df_union=df1.unionByName(df2)
display(df_union)
```

#Date Funcations

```python
from pyspark.sql.functions import current_timestamp
df_date=df_csv.withColumn("date_field",current_timestamp())

display(df_date)
```

```python
from pyspark.sql.functions import date_add

df_add=df_date.withColumn("date_field",date_add("date_field",7))
display(df_add)

```

```python
from pyspark.sql.functions import date_sub
df_sub=df_date.withColumn("date_field",date_sub("date_field",7))

display(df_sub)

```

```python
from pyspark.sql.functions import date_diff

df_diff=df_csv.withColumn("duration",date_diff("order_date","ship_date"))
display(df_diff)

df_select=df_diff.select("order_id","order_date","ship_date","duration")
display(df_select)
```

```python
from pyspark.sql.functions import date_format
df_format=df_date.withColumn("date_field",date_format("date_field","yyyy-MM-dd"))
display(df_format)
```

#String Functions

```python
from pyspark.sql.functions import col,upper,lower
df_string=df_csv.withColumn("order_status",lower(col("order_status")))
display(df_string)
```

```python
from pyspark.sql.functions import length 
df_str=df_csv.withColumn("order_status",length(col("order_status")))

display(df_str)
```
