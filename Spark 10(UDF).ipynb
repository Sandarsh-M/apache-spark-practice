#User Defined Functions

```python
def my_square(x):
    square=x*x
    final=round(square,2)
    return final
```

```python
my_square(20.1)
```

```python
my_square_udf=udf(my_square)
```

```python
df_csv=spark.read.format("csv")\
        .option("header",True)\
        .option("inferSchema",True)\
        .load("/Volumes/sparkcatalog/raw/source/raw_orders/orders.csv")

display(df_csv)
```

```python
df_cust=df_csv.withColumn("price_square",my_square_udf("price"))

display(df_cust)
```

```python
@udf
def my_square_new(x):
    square=x*x
    return square
```

```python
df_cust_new=df_csv.withColumn("price_square",my_square_new("price"))
display(df_cust_new)
```

#User Defined Table Funcations (UDTF)

```python
df_demo = spark.createDataFrame(
    [("Hello World",), ("Just an Example",)],
    ["text"]
)

display(df_demo)
```

```python
from pyspark.sql.functions import udtf
@udtf(returnType="word: string")

class udtf_class:
    def eval(self, text:str):
        for word in text.split():
            yield (word,)
```
