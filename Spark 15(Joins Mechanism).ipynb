```python
data1=[(1,'abc',901),
       (2,'def',902),
       (3,'ghi',903),
       (4,'jkl',904),
       (5,'mno',905),
       (6,'pqr',906),
       (7,'stu',907),
       (8,'vwx',908),
       (9,'yza',909),
       (10,'bcd',910),
       (11,'efg',911),
       (12,'hij',912),
       (13,'klm',913),
       (14,'nop',914),
       (15,'qrs',915),
       (16,'tuv',916),
       (17,'wxy',917),
       (18,'zab',918),
       (19,'cde',919),
       (20,'fgh',920),
       (21,'ijk',921),
       (22,'lmn',922),
       (23,'opq',923),
       (24,'rst',924),
       (25,'uvw',925)]

df1=spark.createDataFrame(data1,['id','name', 'code'])


data2=[(101,1),
       (102,2),
       (103,3),
       (104,4),
       (105,5),
       (106,6),
       (107,7),
       (108,8),
       (109,9),
       (110,10),
       (111,11),
       (112,12),
       (113,13),
       (114,14),
       (115,15),
       (116,16),
       (117,17),
       (118,18),
       (119,19),
       (120,20),
       (121,21),
       (122,22),
       (123,23),
       (124,24),
       (125,25)]

df2=spark.createDataFrame(data2,['o_id','id'])

display(df1)
display(df2)
```

```python
df_inner=df1.join(df2,df1['id']==df2['id'], 'inner')

display(df_inner)
```

###Broadcast Join

```python
df_broadcast=df1.join(broadcast(df2),df1['id']==df2['id'], 'inner')

display(df_broadcast)
```
